Marks of 22 students in 2 subjects

```python
import matplotlib.pyplot as plt

x = [ 5,  4, 13, 11, 13,  5,  3,  4,  3, 10, 12, 13, 14, 12, 11,  6,  7,  6,  5, 11, 14, 10]
y = [15, 18, 22, 21, 21, 19, 15, 16, 16, 21, 23, 22, 22, 20, 22, 17, 18, 14, 17, 24, 22, 23]

plt.scatter(x, y, color='black')
plt.show()
```

```python
data = list(zip(x, y))
print(data)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, n_init='auto')
kmeans.fit(data)

c=kmeans.labels_
print(c)
```

```python
plt.scatter(x, y, c=kmeans.labels_)
plt.show()
```

```python
from sklearn import metrics

metrics.silhouette_score(data, c, metric='euclidean')
```

```python
metrics.calinski_harabasz_score(data, c)
```

```python
metrics.davies_bouldin_score(data, c)
```

```python
from sklearn.cluster import KMeans

data = list(zip(x, y))
inertias = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, n_init='auto')
    kmeans.fit(data)
    inertias.append(kmeans.inertia_)

plt.plot(range(1,11), inertias, marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Sum os Square Error')
print(inertias)
plt.show()
```

```python
from sklearn.cluster import KMeans

data = list(zip(x, y))
inertias = []

for i in range(2,11):
    kmeans = KMeans(n_clusters=i, n_init='auto')
    kmeans.fit(data)
    c=kmeans.labels_
    inertias.append(metrics.silhouette_score(data, c, metric='euclidean'))
print(inertias)
```

```python
plt.plot(range(2,11), inertias, marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('silhouette score')
print(inertias)
plt.show()
```

Clustering Data with Higher Dimensions

```python
import matplotlib.pyplot as plt

x = [4, 5, 10, 4, 3, 11, 14 , 6, 10, 12]
y = [21, 19, 24, 17, 16, 25, 24, 22, 21, 21]
z = [15, 18, 22, 21, 21, 19, 15, 16, 16, 17]

from mpl_toolkits import mplot3d
ax = plt.axes(projection ="3d")

ax.scatter3D(x, y, z, color = "green")
plt.show()
```

```python
from sklearn.cluster import KMeans

data = list(zip(x, y, z))
inertias = []

for i in range(1,11):
    kmeans = KMeans(n_clusters=i, n_init='auto')
    kmeans.fit(data)
    inertias.append(kmeans.inertia_)

plt.plot(range(1,11), inertias, marker='o')
plt.title('Elbow method')
plt.xlabel('Number of clusters')
plt.ylabel('Sum of Square Error')
plt.show()
```

```python
from sklearn.cluster import KMeans

data = list(zip(x, y, z))

kmeans = KMeans(n_clusters=2, n_init='auto')
kmeans.fit(data)

c=kmeans.labels_
print(c)
```

```python
ax = plt.axes(projection ="3d")
ax.scatter3D(x, y, z, c=kmeans.labels_)
plt.show()
```

Cluster Evaluation without Ground Truth
